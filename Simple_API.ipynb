```python
from randomuser import RandomUser
import pandas as pd

r = RandomUser()
def get_users():
  users = []

  for user in r.generate_users(10):
    users.append({
        "Name":user.get_full_name(),
        "Gender":user.get_gender(),
        "City":user.get_city(),
        "State":user.get_state(),
        "Email":user.get_email(),
        "DOB":user.get_dob(),
        "Picture":user.get_picture()
        })

  return pd.DataFrame(users)
```

```python
get_users()
```

```python
df1 = pd.DataFrame(get_users())
```

The requests library.
The requests library is for making HTTP requests in Python.The data is in a json format. We will retrieve results using json.loads() function. json.loads() is used to convert a JSON string into a Python dictionary.
We will convert our json data into pandas data frame.

```python
import requests
import json

data = requests.get("https://fruityvice.com/api/fruit/all")
results = json.loads(data.text)

pd.DataFrame(results)
```

```python
df2 = pd.json_normalize(results)
```

```python
df2
```

use of .iloc[] in pandas

.iloc[] is a pandas indexing method used to select rows and columns by their integer positions (not labels). It stands for "integer-location based indexing"

The syntax is:
    df.iloc[row_index, column_index]

```python
cherry = df2.loc[df2["name"] == 'Cherry']  # Filters rows where name is cherry
```

```python
(cherry.iloc[0]['family']), (cherry.iloc[0]['genus'])
```

```python
banana = df2.loc[df2["name"] == 'Banana']
(banana.iloc[0]['nutritions.calories'])
```
